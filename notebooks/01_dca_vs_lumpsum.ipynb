# **Dollar-Cost Averaging (DCA) vs Lump-Sum Investing**
## **Overview**

This notebook explores the performance differences between two common investment strategies:
1. Lump-sum investing: Investing the entire available capital immediately.
2. Dollar-cost averaging (DCA): Investing capital gradually over time at regular intervals.

The goal is to compare these strategies using historical market data and understand when each approach tends to perform better.
